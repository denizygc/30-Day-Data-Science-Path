## 2. Series vs DataFrame

A **DataFrame** is the whole table. A **Series** is a single column. This distinction matters because they have different methods and behaviors.

````python
# Single brackets -> Series
age_series = df["age"]
print(type(age_series))
print(age_series)
print()

# Double brackets -> DataFrame (with one column)
age_df = df[["age"]]
print(type(age_df))
print(age_df)
````

````python
# A Series has an index and values
print("Index:", age_series.index.tolist())
print("Values:", age_series.values)
print("Name:", age_series.name)
print("Dtype:", age_series.dtype)
````

````python
# Series has methods that DataFrames don't (and vice versa)
# .str accessor only works on a Series of strings
print(df["name"].str.upper())
print()

# You can do math directly on a numeric Series
print("Mean age:", df["age"].mean())
print("Total spending:", df["spending"].sum())
````

## 5. Filtering with Boolean Indexing

This is the most common way to select rows. You write a condition, and pandas returns only the rows where it's True.

````python
# Step 1: The condition creates a boolean Series (True/False for each row)
mask = df["age"] > 30
print(mask)
print(f"\nType: {type(mask)}")
````

````python
# Step 2: Pass that mask into the brackets to filter
older_than_30 = df[df["age"] > 30]
print(f"People older than 30: {len(older_than_30)}")
older_than_30
````

````python
# Combining conditions: use & (and), | (or), ~ (not)
# IMPORTANT: wrap each condition in parentheses!

# People older than 30 AND spending more than 300
high_value = df[(df["age"] > 30) & (df["spending"] > 300)]
print("High-value customers:")
print(f"\nHigh-value customers: {len(high_value)} rows")
print(high_value[["name", "age", "spending"]])
print()

# People from Berlin OR Ankara
not_istanbul = df[(df["city"] == "Berlin") | (df["city"] == "Ankara")]
print("Berlin or Ankara:")
print(f"\nBerlin or Ankara: {len(not_istanbul)} rows")
print(not_istanbul[["name", "city"]])
````

````python
# Shortcut: .isin() is cleaner than chaining multiple | conditions
istanbul_berlin = df[df["city"].isin(["Istanbul", "Berlin"])]
print(f"Istanbul or Berlin: {len(istanbul_berlin)} rows")

# Negate with ~ (tilde) = NOT
not_berlin = df[~(df["city"] == "Berlin")]
print(f"\nNot Berlin: {len(not_berlin)} rows")
print(not_berlin[["name", "city"]])
````

````python
# String filtering with .str accessor
# Names that start with a vowel (case-insensitive)
vowel_names = df[df["name"].str.lower().str[1].isin([ "l"])]
print("Names starting with a vowel:")
print(vowel_names["name"].tolist())

# Names containing "an"
has_an = df[df["name"].str.contains("an", case=False)]
print("\nNames containing 'an':")
print(has_an["name"].tolist())
````

## 6. loc and iloc

These are the two main ways to select specific rows AND columns at the same time.

| | **loc** | **iloc** |
|---|---|---|
| **Selects by** | Labels (names) | Integer position |
| **End inclusive?** | Yes | No |
| **Syntax** | `df.loc[rows, cols]` | `df.iloc[rows, cols]` |

````python
# loc: select by LABEL
# Rows 0 through 3 (INCLUSIVE), columns "name" and "city"
df.loc[0:3, ["name", "city"]]
````

````python
# iloc: select by POSITION (like Python list slicing)
# Rows at position 0, 1, 2 (NOT 3), columns at position 0 and 1
df.iloc[0:3, 0:2]
````

````python
# THE BIG GOTCHA: loc includes the endpoint, iloc doesn't
# Both start at row 0 and go to "4"

loc_result = df.loc[0:4, "name"]    # 5 rows (0,1,2,3,4)
iloc_result = df.iloc[0:4, 0]        # 4 rows (0,1,2,3)

print(f"loc[0:4]  -> {len(loc_result)} rows: {loc_result.tolist()}")
print(f"iloc[0:4] -> {len(iloc_result)} rows: {iloc_result.tolist()}")
````

````python
# loc with a boolean mask (very common pattern)
# "Give me the name and spending of people older than 30"
df.loc[df["age"] > 30, ["name", "spending"]]
````

````python
# loc to MODIFY values (the safe way to assign)
df_copy = df.copy()

# Set spending to 999 for everyone in Berlin
df_copy.loc[df_copy["city"] == "Ankara", "spending"] = 999
print(df_copy[["name", "city", "spending"]])
````

## 7. Groupby Basics

`groupby` splits your data into groups, applies a function to each group, and combines the results. It's like SQL's `GROUP BY`.

````python
# Basic groupby: average spending per city
df.groupby("city")["spending"].mean()
````

````python
# Multiple aggregations with .agg()
city_stats = df.groupby("city")["spending"].agg(["mean", "sum", "count", "min", "max"])
city_stats
````

````python
# Different aggregations for different columns
multi_agg = df.groupby("city").agg(
    avg_age=("age", "mean"),
    total_spending=("spending", "sum"),
    num_customers=("name", "count")
)
multi_agg
````

````python
# Group by multiple columns
df.groupby(["city", "gender"])["spending"].mean()
````

````python
# What does the GroupBy object actually look like inside?
# .groups shows you which rows belong to which group
grouped = df.groupby("city")
print("Groups:", grouped.groups)
print()

# You can iterate over groups
for group_name, group_df in grouped:
    print(f"--- {group_name} ({len(group_df)} rows) ---")
    print(group_df[["name", "spending"]].to_string(index=False))
    print()
````

## 9. Tricky Bits and Common Mistakes

Things that trip up beginners (and sometimes experienced people too).

````python
# GOTCHA 1: Chained indexing creates a copy, not a view
# This might NOT change your original DataFrame
df_test = df.copy()
print(df_test)

# BAD: chained indexing
# df_test["spending"][0] = 9999  # SettingWithCopyWarning!

# GOOD: use loc
df_test.loc[9, "spending"] = 9999
print(f"Row 9 spending is now: {df_test.loc[9, 'spending']}")
print(df_test)
````

````python
# GOTCHA 2: df2 = df does NOT copy the data
df_a = pd.DataFrame({"x": [1, 2, 3]})
df_b = df_a           # SAME object, not a copy
df_c = df_a.copy()    # INDEPENDENT copy

df_b.loc[0, "x"] = 999

print(f"df_a[0] = {df_a.loc[0, 'x']}  (changed because df_b is the same object)")
print(f"df_c[0] = {df_c.loc[0, 'x']}  (unchanged because df_c is a copy)")
````

````python
# GOTCHA 3: Forgetting parentheses in compound filters
# This FAILS, but the notebook keeps running because we catch any exception:
try:
    result = df[df["age"] > 30 & df["spending"] > 300]
    print("Wrong filter unexpectedly worked:")
except Exception as e:
    print(f"Error: {e}")
    print("Fix: wrap each condition in parentheses")
    print()

# This WORKS:
result = df[(df["age"] > 30) & (df["spending"] > 300)]
print("Correct result:")
print(result[["name", "age", "spending"]])
````

````python
# GOTCHA 4: groupby without aggregation gives you an object, not results
grouped = df.groupby("city")
print(type(grouped))  # DataFrameGroupBy, not a DataFrame!
print(grouped)        # Just shows the object, no data
print()

# You MUST call an aggregation
print("After .mean():")
print(grouped["spending"].mean())
````

````python
# GOTCHA 5: read_csv guesses dtypes wrong sometimes
# Zip codes, IDs, phone numbers that start with 0 get read as integers
import io

csv_data = """name,zipcode
Ali,00123
Bea,10115"""

# BAD: pandas drops the leading zeros
df_bad = pd.read_csv(io.StringIO(csv_data))
print("Without dtype fix:")
print(df_bad)
print(f"  zipcode type: {df_bad['zipcode'].dtype}")
print()

# GOOD: force string type
df_good = pd.read_csv(io.StringIO(csv_data), dtype={"zipcode": str})
print("With dtype={'zipcode': str}:")
print(df_good)
print(f"  zipcode type: {df_good['zipcode'].dtype}")
````

## 10. Trick Questions

Think about each one before revealing the answer.

<details>
<summary><strong>Q1:</strong> What does <code>df.loc[3:5]</code> return vs <code>df.iloc[3:5]</code>?</summary>

`loc[3:5]` returns rows with index labels 3, 4, and 5 (3 rows, both ends inclusive). `iloc[3:5]` returns rows at positions 3 and 4 (2 rows, end exclusive).
</details>

<details>
<summary><strong>Q2:</strong> You run <code>df[df["score"] != np.nan]</code> and get ALL rows back, even though some scores are missing. Why?</summary>

Any comparison with NaN returns False or True in unexpected ways. `NaN != NaN` is True, so every row passes the filter. Use `df[df["score"].notna()]` instead.
</details>

<details>
<summary><strong>Q3:</strong> You do <code>df2 = df</code> and then modify df2. Does df change too?</summary>

Yes! `df2 = df` makes both variables point to the same object. Use `df2 = df.copy()` to get an independent copy.
</details>

<details>
<summary><strong>Q4:</strong> Is <code>df["name"]</code> a DataFrame or a Series?</summary>

It's a Series. Single brackets with one column name always return a Series. Use `df[["name"]]` (double brackets) to get a DataFrame with one column.
</details>

<details>
<summary><strong>Q5:</strong> What happens if you call <code>df.groupby("city")</code> without an aggregation?</summary>

You get a DataFrameGroupBy object, not actual results. You must call `.mean()`, `.sum()`, `.agg()`, etc. to see numbers.
</details>
